```python
import os
import pandas as pd

# ============================================================
# PASTA ATUAL (mesma do notebook)
# ============================================================

CSV_FOLDER = "."

POSSIBLE_TARGETS = [
    "target",
    "label",
    "class",
    "y",
    "categoria",
    "classe"
]

results = []

csv_files = [
    f for f in os.listdir(CSV_FOLDER)
    if f.endswith(".csv")
]

print(f"\n📂 Total CSVs encontrados: {len(csv_files)}\n")

for file in csv_files:

    print(f"🔍 Analisando: {file}")

    try:

        df = pd.read_csv(file)

        target_col = None

        # tenta detectar target
        for col in POSSIBLE_TARGETS:

            if col in df.columns:

                target_col = col
                break

        # fallback = última coluna
        if target_col is None:

            target_col = df.columns[-1]

            print(
                f"   ⚠️ Usando última coluna: {target_col}"
            )

        n_classes = df[target_col].nunique()

        results.append({
            "arquivo": file,
            "target": target_col,
            "n_classes": n_classes,
            "n_samples": len(df),
            "multiclass": n_classes > 2
        })

        print(f"   ✅ Classes: {n_classes}")

    except Exception as e:

        print(f"   ❌ Erro: {e}")

# ============================================================
# RESULTADOS
# ============================================================

results_df = pd.DataFrame(results)

# ============================================================
# DATASETS COM 10 A 50 CLASSES (INCLUSIVO)
# ============================================================

datasets_10_50 = results_df[
    (results_df["n_classes"] >= 10) &
    (results_df["n_classes"] <= 5000000000000000000000000000000000000000000000)
]

print("\n================ 10 A 50 CLASSES ================\n")

print(
    datasets_10_50[
        ["arquivo", "n_classes"]
    ].sort_values(
        by="n_classes",
        ascending=False
    )
)

print(
    f"\n📊 Total de datasets com entre 10 e 50 classes: "
    f"{len(datasets_10_50)}"
)

print("\n================ RESULTADOS ================\n")

print(
    results_df.sort_values(
        by="n_classes",
        ascending=False
    )
)

results_df.to_csv(
    "datasets_num_classes.csv",
    index=False
)

print(
    "\n💾 Resultado salvo em datasets_num_classes.csv"
)
```
