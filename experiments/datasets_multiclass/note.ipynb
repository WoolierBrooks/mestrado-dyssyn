```python
import os
import pandas as pd

# ============================================================
# PASTA ATUAL (mesma do notebook)
# ============================================================

CSV_FOLDER = "."

POSSIBLE_TARGETS = [
    "target",
    "label",
    "class",
    "y",
    "categoria",
    "classe"
]

results = []

csv_files = [
    f for f in os.listdir(CSV_FOLDER)
    if f.endswith(".csv")
]

print(f"\n📂 Total CSVs encontrados: {len(csv_files)}\n")

for file in csv_files:

    print(f"🔍 Analisando: {file}")

    try:

        df = pd.read_csv(file)

        target_col = None

        # Tenta detectar target
        for col in POSSIBLE_TARGETS:
            if col in df.columns:
                target_col = col
                break

        # Fallback = última coluna
        if target_col is None:
            target_col = df.columns[-1]
            print(f"   ⚠️ Usando última coluna: {target_col}")

        n_classes = df[target_col].nunique()
        n_linhas = len(df)

        results.append({
            "arquivo": file,
            "target": target_col,
            "n_linhas": n_linhas,
            "n_classes": n_classes,
            "multiclass": n_classes > 2
        })

        print(f"   ✅ Linhas: {n_linhas:,} | Classes: {n_classes}")

    except Exception as e:

        print(f"   ❌ Erro: {e}")

# ============================================================
# RESULTADOS
# ============================================================

results_df = pd.DataFrame(results)

if not results_df.empty:

    # ============================================================
    # DATASETS COM 10 A 50 CLASSES (ORDENADO POR MENOR NÚMERO DE LINHAS)
    # ============================================================

    datasets_10_50 = results_df[
        (results_df["n_classes"] >= 10) &
        (results_df["n_classes"] <= 50)
    ].sort_values(by="n_linhas", ascending=True)

    print("\n================ 10 A 50 CLASSES (MENOR → MAIOR EM LINHAS) ================\n")

    print(
        datasets_10_50[
            ["arquivo", "n_linhas", "n_classes"]
        ]
    )

    print(
        f"\n📊 Total de datasets com entre 10 e 50 classes: "
        f"{len(datasets_10_50)}"
    )

    # ============================================================
    # RESULTADOS GERAIS (ORDENADO POR MENOR NÚMERO DE LINHAS)
    # ============================================================

    results_df = results_df.sort_values(by="n_linhas", ascending=True)

    print("\n================ RESULTADOS GERAIS (MENOR → MAIOR EM LINHAS) ================\n")

    print(
        results_df[
            ["arquivo", "n_linhas", "n_classes", "target", "multiclass"]
        ]
    )

    results_df.to_csv(
        "datasets_num_classes.csv",
        index=False
    )

    print(
        "\n💾 Resultado salvo em datasets_num_classes.csv"
    )
```
